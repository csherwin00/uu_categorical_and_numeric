```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('../data/metabric.csv')
```

```python
sns.ecdfplot(data=df, x='Overall Survival (Months)')
plt.show()

df['Overall Survival (Months)'].describe()
```

```python
sns.ecdfplot(data=df, x='Overall Survival (Months)', hue='Chemotherapy')
plt.show()

df.groupby('Chemotherapy')['Overall Survival (Months)'].median()
```

```python
cols = ['Age at Diagnosis', 'Lymph nodes examined positive',
        'Tumor Size', 'Tumor Stage', 'Nottingham prognostic index']
df.groupby('Chemotherapy')[cols].mean()
```

```python
cols = ['Age at Diagnosis', 'Lymph nodes examined positive',
        'Tumor Size', 'Tumor Stage', 'Nottingham prognostic index']
df.groupby('Chemotherapy')[cols].mean()
sns.ecdfplot(data=df, x='Nottingham prognostic index', hue='Chemotherapy')
plt.show()
```
